This joins the current runtime to a MoreGPU pool as a worker. It computes on the runtime's GPU
(if present) or CPU, and adapts its duty cycle to the machine's load.

```python
# 1. Configure — paste the values from your admin server's setup wizard.
MOREGPU_SERVER = "wss://YOUR_ADMIN_HOST:8787/ws"   # your pool's admin server (use wss:// in production)
MOREGPU_TOKEN  = "YOUR_JOIN_TOKEN"                 # the pool's join token
MAX_UTIL       = "0.85"                            # keep total machine utilization under this

assert "YOUR_ADMIN_HOST" not in MOREGPU_SERVER and MOREGPU_TOKEN != "YOUR_JOIN_TOKEN", \
    "Fill in MOREGPU_SERVER and MOREGPU_TOKEN first."
```

```python
# 2. Install Deno (single binary) if needed.
import os, shutil
if shutil.which('deno') is None:
    os.system('curl -fsSL https://deno.land/install.sh | sh')
os.environ['PATH'] = os.path.expanduser('~/.deno/bin') + ':' + os.environ['PATH']
print('deno:', shutil.which('deno') or 'not found')
```

```python
# 3. Join the pool. This cell runs until you stop it (the worker stays connected).
import os
os.environ.update(MOREGPU_SERVER=MOREGPU_SERVER, MOREGPU_TOKEN=MOREGPU_TOKEN, MOREGPU_MAX_UTIL=MAX_UTIL)
worker = 'https://raw.githubusercontent.com/ArioMoniri/moregpu/main/apps/worker/worker.ts'
!deno run --unstable-webgpu --allow-net --allow-env --allow-sys "$worker" \
  --server "$MOREGPU_SERVER" --token "$MOREGPU_TOKEN"
```
